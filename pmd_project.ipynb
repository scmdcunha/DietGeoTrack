# DOWNLOAD E EXPLORAÇÃO DO DATASET


Este trabalho utiliza o dataset **A Pressure Map Dataset for In-bed Posture Classification** disponível no PhysioNet, que contém medições de pressão corporal obtidas através de um colchão equipado com sensores.

Cada ficheiro corresponde a **uma postura específica de um sujeito**, sendo os dados representados como uma matriz de valores de pressão (em mmHg), onde:

- Cada **linha** corresponde a um instante temporal (**frame**)
- Cada **coluna** corresponde a um **sensor de pressão**
- O sistema possui aproximadamente 2048 sensores (32×64)
- Valores iguais a 0 indicam **ausência de contacto corporal**

O dataset está dividido em dois experimentos:

- **Experiment I**: diferentes posturas em condições controladas
- **Experiment II**: comparação entre dois tipos de colchão
  - Air mattress
  - Sponge mattress

O objetivo da análise é caracterizar a distribuição de pressão através de métricas como:

- Número de sensores ativos (contact area)
- Pressão média (global e em sensores ativos)
- Pressão máxima
- Comparação entre tipos de colchão
